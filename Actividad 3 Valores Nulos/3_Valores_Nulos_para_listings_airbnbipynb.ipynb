**Métodos de Sustitución de Valores Nulos**


```python
#Realizamos una copia del dataframe
data2=data.copy()
```

```python
data2.info()
```

```python
#Primer método de sustitución de valores nulos
#Sustituir valores nulos con promedio o media
# Crear copia para aplicar sustitución de valores nulos
data2 = data.copy()

# Sustituir valores nulos de una columna categórica con la moda
moda_host_location = data2["host_location"].mode().iloc[0]
data2["host_location"] = data2["host_location"].fillna(moda_host_location)
```

```python
#Corroboramos valores nulos
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
#Segundo método de sustitución de valores nulos
#Sustituir valores nulos con mediana
data2["bathrooms"] = data2["bathrooms"].fillna(round(data2["bathrooms"].median(), 1))
```

```python
#Segundo método de sustitución de valores nulos
#Sustituir valores nulos con mediana
data2["price_quote_total_price"] = data2["price_quote_total_price"].fillna(round(data2["price_quote_total_price"].median(), 1))
```

```python

#Segundo método de sustitución de valores nulos
#Sustituir valores nulos con mediana
data2["price_quote_price_per_night"] = data2["price_quote_price_per_night"].fillna(round(data2["price_quote_price_per_night"].median(), 1))
```

```python
#Corroboramos valores nulos
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
#Corroboramos valores nulos
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
#Tercer método de sustitución de valores nulos
#Sustituir valores nulos por un valor numérico en  concreto 
data2["bedrooms"]= data2["bedrooms"].fillna(0) 
data2
```

```python
#Corroboramos valores nulos
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
data2["beds"] = data2["beds"].fillna( data2["beds"].median())
data2
```

```python
#Corroboramos valores nulos
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
#Cuarto método de sustitución de valores nulos
#Sustituir valores nulos por un string en  concreto 
data2["price_quote_checkin_date"] =data2["price_quote_checkin_date"].fillna("Sin fecha") 
```

```python
#Cuarto método de sustitución de valores nulos
#Sustituir valores nulos por un string en  concreto 
data2["license"] =data2["license"].fillna("Sin licencia") 

```
