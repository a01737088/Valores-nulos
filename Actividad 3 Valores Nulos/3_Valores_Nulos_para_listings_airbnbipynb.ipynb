```python
# JUSTIFICACION DEL TRATAMIENTO DE VALORES NULOS

# host_location:
# Se utilizo la moda porque es una variable categorica y permite sustituir
# los nulos con la ubicacion que aparece con mayor frecuencia.

# bathrooms:
# Se utilizo la mediana porque es una variable numerica y este metodo
# evita que los alojamientos con muchos baños alteren la sustitucion.

# price_quote_total_price:
# Se utilizo la mediana porque existen precios extremos que pueden afectar el promedio.

# price_quote_price_per_night:
# Se utilizo la mediana porque los precios por noche presentan valores extremos.

# bedrooms:
# Se utilizo el valor numerico 0 para representar posibles alojamientos tipo estudio.

# beds:
# Se utilizo la mediana porque es una variable numerica y puede presentar
# alojamientos con cantidades de camas muy altas.

# price_quote_checkin_date:
# Se utilizo el string "Sin fecha" para identificar los registros
# que no contaban con una fecha de entrada.

# price_quote_checkout_date:
# Se utilizo el string "Sin fecha" para identificar los registros
# que no contaban con una fecha de salida.

# license:
# Se utilizo el string "Sin licencia" para conservar los alojamientos
# que no tenian una licencia registrada.

# reviews_per_month:
# Se utilizo el valor numerico 0 para representar alojamientos
# que no registraron reseñas durante el mes.

# En las variables restantes que presentaban valores nulos se eliminaron
# las filas incompletas mediante dropna(), porque no se aplico otro metodo
# de sustitucion y se necesitaba generar un archivo final sin valores nulos.

# Las demas variables no necesitaron tratamiento porque originalmente
# no presentaban valores nulos.
```

**Métodos de Sustitución de Valores Nulos**


```python
#Realizamos una copia del dataframe
data2=data.copy()
```

```python
data2.info()
```

```python
#Primer método de sustitución de valores nulos
#Sustituir valores nulos con promedio o media
# Crear copia para aplicar sustitución de valores nulos
data2 = data.copy()

# Sustituir valores nulos de una columna categórica con la moda
moda_host_location = data2["host_location"].mode().iloc[0]
data2["host_location"] = data2["host_location"].fillna(moda_host_location)
```

```python
#Corroboramos valores nulos
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
#Segundo método de sustitución de valores nulos
#Sustituir valores nulos con mediana
data2["bathrooms"] = data2["bathrooms"].fillna(round(data2["bathrooms"].median(), 1))
```

```python
#Segundo método de sustitución de valores nulos
#Sustituir valores nulos con mediana
data2["price_quote_total_price"] = data2["price_quote_total_price"].fillna(round(data2["price_quote_total_price"].median(), 1))
```

```python

#Segundo método de sustitución de valores nulos
#Sustituir valores nulos con mediana
data2["price_quote_price_per_night"] = data2["price_quote_price_per_night"].fillna(round(data2["price_quote_price_per_night"].median(), 1))
```

```python
#Corroboramos valores nulos
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
#Corroboramos valores nulos
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
#Tercer método de sustitución de valores nulos
#Sustituir valores nulos por un valor numérico en  concreto 
data2["bedrooms"]= data2["bedrooms"].fillna(0) 
data2
```

```python
#Corroboramos valores nulos
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
data2["beds"] = data2["beds"].fillna( data2["beds"].median())
data2
```

```python
#Corroboramos valores nulos
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
#Cuarto método de sustitución de valores nulos
#Sustituir valores nulos por un string en  concreto 
data2["price_quote_checkin_date"] =data2["price_quote_checkin_date"].fillna("Sin fecha") 
```

```python
#Cuarto método de sustitución de valores nulos
#Sustituir valores nulos por un string en  concreto 
data2["license"] =data2["license"].fillna("Sin licencia") 

```

```python
#Corroboramos valores nulos
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
#Cuarto método de sustitución de valores nulos
#Sustituir valores nulos por un string en  concreto 
data2["price_quote_checkout_date"] =data2["price_quote_checkout_date"].fillna("Sin fecha") 
```

```python
#Corroboramos valores nulos
valores_nulos=data2.isnull().sum()
valores_nulos
```
